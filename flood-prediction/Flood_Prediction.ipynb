```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.preprocessing import LabelEncoder
```

```python
f = pd.read_csv('C:\\Dev\\ml-projects\\flood-prediction\\flood.csv')
```

```python
f.head()
```

```python
f.describe()
```

```python
f.info()
```

```python
# vlaue counts 
f['Flood'] = (f['FloodProbability'] > 0.5).astype(int)
print(f['Flood'].value_counts())
```

```python
f.head()
```

```python
x = f.drop(columns=['Flood', 'FloodProbability'], axis=1)
y = f['Flood']
```

```python
# Splitting into target variables
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=43)
```

```python
print(x_train.shape, x_test.shape, y_train.shape, y_test.shape)
```

```python
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print()
print(classification_report(y_test, y_pred))
```

```python
fig, ax = plt.subplots(figsize=(14, 8))
fig.patch.set_facecolor('#080808')
ax.set_facecolor('#0f0505')
corr = f.drop(columns=['Flood']).corr()
sns.heatmap(corr, ax=ax, cmap='Blues', center=0,
            linewidths=0.5, linecolor='#080808',
            annot=False, square=True)
ax.tick_params(colors='#9a7070', labelsize=8)
plt.tight_layout()
plt.show()
```

```python
feature_imp = pd.Series(model.feature_importances_, index=x.columns).sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(10, 6))
fig.patch.set_facecolor('#080808')
ax.set_facecolor('#0d1a35')

top10 = feature_imp.head(10)
colors = ['#00e5ff' if i == 0 else '#4fc3f7' if i < 3 else '#0288d1' for i in range(len(top10))]
ax.barh(top10.index[::-1], top10.values[::-1], color=colors[::-1], height=0.55)

ax.set_xlabel('Importance Score', color='#5c7aaa', fontsize=9)
ax.tick_params(colors='#8ab4d4', labelsize=9)
for spine in ax.spines.values():
    spine.set_edgecolor('#1e3a5f')

for i, (val, name) in enumerate(zip(top10.values[::-1], top10.index[::-1])):
    ax.text(val + 0.0003, i, f'{val:.4f}', va='center', color='#8ab4d4', fontsize=8)

plt.title('Feature Importance Rankings — Top 10 Predictor Variables', color='#8ab4d4', fontsize=11)
plt.tight_layout()
plt.show()
```
