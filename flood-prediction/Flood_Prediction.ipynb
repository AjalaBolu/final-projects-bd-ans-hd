```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.preprocessing import LabelEncoder
```

```python
f = pd.read_csv('/content/flood.csv')
```

```python
f.head()
```

```python
f.describe()
```

```python
f.info()
```

```python
# Turning the flood coloums into
f['Flood'] = (f['FloodProbability'] > 0.5).astype(int)
print(f['Flood'].value_counts())
```

```python
f.head()
```

```python
x = f.drop(columns=['Flood', 'FloodProbability'], axis=1)
y = f['Flood']
```

```python
# Splitting into target variables
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=43)
```

```python
print(x_train.shape, x_test.shape, y_train.shape, y_test.shape)
```

```python
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print()
print(classification_report(y_test, y_pred))
```
